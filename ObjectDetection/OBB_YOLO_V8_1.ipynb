# Brain OBB Feature Detection Training

This notebook trains a YOLOv8 Oriented Bounding Box (OBB) model to detect features in brain images.  
It assumes you have a dataset of images and OBB annotations.

```python
import os
import ultralytics
from ultralytics import YOLO
from IPython import display

HOME = os.getcwd()
print(f"Current working directory: {HOME}")

# Check YOLO installation
ultralytics.checks()
```

## 1. Prepare Data Splits

We use `prepare_splits.py` to organize our `dataset/brain/images` and `dataset/brain/annotations` into Train and Validation sets required by YOLO.

```python
# Make sure we are in the project root
%cd "d:\Desktop\6th Sem\CV\Project"

# Run data preparation script
!python prepare_splits.py --dataset dataset/brain
```

## 2. Train Model

We use `yolov8n-obb.pt` (Nano) as the base model for speed. You can start with `yolov8m-obb.pt` for better accuracy if you have a good GPU.
Training uses `brain_config.yaml` to know where the data classes are.

```python
# Load a pretrained model
model = YOLO("yolov8n-obb.pt")

# Train the model
# epochs=50: Train for 50 passes over the dataset
# imgsz=640: Resize images to 640x640
# device=0: Use GPU 0
results = model.train(data="brain_config.yaml", epochs=50, imgsz=640)

print("Training complete.")
```

## 3. Validation and Export

Check how well the model performs and export it for use.

```python
# Evaluate performance on validation set
metrics = model.val()
print(f"mAP50: {metrics.box.map50}")
print(f"mAP50-95: {metrics.box.map}")

# Export to ONNX (optional, good for deployment)
success = model.export(format="onnx")
print(f"Export success: {success}")
```
